```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(pulsep, pulsen, fth, c=fth, cmap="viridis", alpha=0.7)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

xi = np.linspace(pulsep.min(), pulsep.max(), 100)
yi = np.linspace(pulsen.min(), pulsen.max(), 100)
X, Y = np.meshgrid(xi, yi)
Z = griddata((pulsep, pulsen), fth, (X, Y), method="cubic")

# replace the matplotlib fig/ax block with this
fig = go.Figure(data=[go.Surface(x=X, y=Y, z=Z)])
fig.show()
```
